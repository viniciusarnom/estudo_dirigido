```python
import numpy as np
import matplotlib.pyplot as plt

# Usando o novo arquivo que você enviou: WiFi_Day_1_bes_s2.bin
data = np.fromfile('WiFi_Day_1_bes_s3.bin', dtype=np.complex64, count=5000)

I = np.real(data)
Q = np.imag(data)

# Criar o gráfico
plt.figure(figsize=(14, 6))
plt.plot(I[:1000], label='Parte I (In-phase)', color='blue')
plt.plot(Q[:1000], label='Parte Q (Quadrature)', color='red', alpha=0.7)
plt.title('Gráfico das Partes I e Q do Sinal (Primeiras 1000 amostras) - Arquivo s3')
plt.xlabel('Índice da Amostra')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig('iq_components_plot_s3.png')
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Carregar os dados do novo arquivo s3
data = np.fromfile('WiFi_Day_1_bes_s3.bin', dtype=np.complex64, count=5000)
x_t = data

# 2. Calcular a magnitude |x(t)|
magnitude_x_t = np.abs(x_t)

# 3. Calcular o espectro X(f) = F{x(t)}
X_f = np.fft.fft(x_t)

# Centralizar a frequência 0 e criar o eixo de frequências (fs = 5 MHz)
fs = 5e6
X_f_shifted = np.fft.fftshift(X_f)
magnitude_X_f = np.abs(X_f_shifted)
freqs = np.fft.fftshift(np.fft.fftfreq(len(x_t), 1/fs))

# 4. Plotar os resultados
plt.figure(figsize=(14, 10))

# Gráfico da Magnitude no tempo |x(t)|
plt.subplot(2, 1, 1)
plt.plot(magnitude_x_t[:1000], color='green')
plt.title('Magnitude do Sinal no Tempo |x(t)| = $\sqrt{I^2 + Q^2}$ (Primeiras 1000 amostras) - Arquivo s3')
plt.xlabel('Índice da Amostra')
plt.ylabel('Magnitude |x(t)|')
plt.grid(True)

# Gráfico do Espectro |X(f)|
plt.subplot(2, 1, 2)
plt.plot(freqs / 1e6, magnitude_X_f, color='purple')
plt.title('Espectro de Frequência |X(f)| = $\mathcal{F}\{x(t)\}$ - Arquivo s3')
plt.xlabel('Frequência (MHz)')
plt.ylabel('Magnitude |X(f)|')
plt.grid(True)

plt.tight_layout()
plt.savefig('magnitude_and_spectrum_s3.png')
```
